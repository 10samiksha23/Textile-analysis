```python
from langchain.chat_models import ChatOpenAI
from langchain.chains.question_answering import load_qa_chain

def ask_question(vectorstore, question):
    # Step 1: Find matching documents using similarity search
    matching_docs = vectorstore.similarity_search(question)

    # Step 2: Load OpenAI GPT model
    llm = ChatOpenAI(temperature=0)

    # Step 3: Load Question-Answering Chain
    qa_chain = load_qa_chain(llm, chain_type="stuff")

    # Step 4: Run the chain with the matching documents
    response = qa_chain.run(input_documents=matching_docs, question=question)

    return response

# ✅ Example
question = "What is Generative AI?"
response = ask_question(vs, question)
print("🤖 Answer:", response)
```
